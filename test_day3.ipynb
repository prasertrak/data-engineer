```python
def write_report(dictionary, report_file):
      with open(report_file, "w+") as f:
          for k in sorted(dictionary):
              f.write(str(k)+':'+str(dictionary[k])+'\n')
              f.close()
              write_report(dictionary, '<report_file>')
```

```python
'''
add error handling
'''
#!/usr/bin/env python3
import csv
import sys

def read_employees(csv_file_location):
    # ลงทะเบียนรูปแบบการอ่านไฟล์ CSV
    csv.register_dialect('empDialect', skipinitialspace=True, strict=True)
    employee_list = []

    try:
        # ใช้ 'with open' เพื่อให้ Python ปิดไฟล์ให้อัตโนมัติเมื่อทำงานเสร็จ (ปลอดภัยกว่าเดิม)
        with open(csv_file_location, mode='r', encoding='utf-8') as file:
            employee_file = csv.DictReader(file, dialect='empDialect')
            for data in employee_file:
                employee_list.append(data)

    except FileNotFoundError:
        print(f"❌ ผิดพลาด: ไม่พบไฟล์ในตำแหน่งที่ระบุ -> '{csv_file_location}'", file=sys.stderr)
        return None
    except PermissionError:
        print(f"❌ ผิดพลาด: ไม่มีสิทธิ์เข้าถึงไฟล์นี้ -> '{csv_file_location}'", file=sys.stderr)
        return None
    except csv.Error as e:
        print(f"❌ ผิดพลาด: รูปแบบไฟล์ CSV ไม่ถูกต้อง -> {e}", file=sys.stderr)
        return None
    except Exception as e:
        print(f"❌ เกิดข้อผิดพลาดที่ไม่คาดคิด: {e}", file=sys.stderr)
        return None

    return employee_list

# เรียกใช้งานฟังก์ชัน
path = '/home/<username>/data/employees.csv'
employee_list = read_employees(path)

# ตรวจสอบผลลัพธ์ก่อนพิมพ์
if employee_list is not None:
    print(f"ดึงข้อมูลสำเร็จ! พบพนักงานทั้งหมด {len(employee_list)} คน:")
    print(employee_list)
else:
    print("ไม่สามารถประมวลผลข้อมูลได้เนื่องจากเกิดข้อผิดพลาดข้างต้น")
```

```python
'''
chang csv lib. to pandas
'''
#!/usr/bin/env python3
# import csv
import pandas as pd
import sys

def read_employees(csv_file_location):
    try:
        # อ่านไฟล์ CSV ด้วย Pandas
        # sep=',' คั่นด้วยคอมมา, skipinitialspace=True ตัดช่องว่างหน้าข้อมูล
        df = pd.read_csv(csv_file_location, sep=',', skipinitialspace=True)

        # แปลง DataFrame ให้กลายเป็น List of Dictionaries เหมือนโค้ดเดิม
        employee_list = df.to_dict(orient='records')
        return employee_list

    except FileNotFoundError:
        print(f"❌ ผิดพลาด: ไม่พบไฟล์ในตำแหน่งที่ระบุ -> '{csv_file_location}'", file=sys.stderr)
        return None
    except PermissionError:
        print(f"❌ ผิดพลาด: ไม่มีสิทธิ์เข้าถึงไฟล์นี้ -> '{csv_file_location}'", file=sys.stderr)
        return None
    except pd.errors.EmptyDataError:
        print(f"❌ ผิดพลาด: ไฟล์ CSV นี้เป็นไฟล์ว่างเปล่า", file=sys.stderr)
        return None
    except Exception as e:
        print(f"❌ เกิดข้อผิดพลาดที่ไม่คาดคิด: {e}", file=sys.stderr)
        return None

# เรียกใช้งานฟังก์ชัน
path = '/home/<username>/data/employees.csv'
employee_list = read_employees(path)

# ตรวจสอบผลลัพธ์ก่อนพิมพ์
if employee_list is not None:
    print(f"ดึงข้อมูลสำเร็จ! พบพนักงานทั้งหมด {len(employee_list)} คน:")
    print(employee_list)
```
