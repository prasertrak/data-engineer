```python
'''
insert table
'''

# 1. Establish the connection to the MySQL database
connection = mysql.connector.connect(
    host="localhost", user="root", password="", database="test"
)

# 2. Create a cursor object to execute SQL statements
cursor = connection.cursor()

# 3. Define the SQL INSERT query with placeholders (%s)
# Replace 'users' with your table name, and 'name', 'email' with your column names
sql_query = "INSERT INTO users (name, email) VALUES (%s, %s)"

# 4. Define the actual data you want to insert into the table
data_to_insert = ("John Doe", "john.doe@example.com")

try:
    # 5. Execute the query with the provided data
    cursor.execute(sql_query, data_to_insert)

    # 6. CRITICAL STEP: Commit the transaction to save changes permanently to the database
    connection.commit()

    # Print success message along with the auto-incremented ID
    print(f"Data inserted successfully. Target ID: {cursor.lastrowid}")

except mysql.connector.Error as error:
    # Handle any errors that occur during the process (e.g., table not found, wrong data type)
    print(f"Failed to insert record into MySQL table: {error}")

finally:
    # 7. Close the cursor and connection to free up system resources
    cursor.close()
    connection.close()
    print("MySQL connection is closed.")
```

```python
'''
select data
'''

import mysql.connector

# 1. Establish the connection to the MySQL database
connection = mysql.connector.connect(
    host="localhost", user="root", password="", database="test"
)

# 2. Create a cursor object
cursor = connection.cursor()

# 3. Define the SQL SELECT query
# Replace 'users' with your actual table name
sql_query = "SELECT id, name, email FROM users"

try:
    # 4. Execute the SELECT statement
    cursor.execute(sql_query)

    # 5. Fetch all rows returned by the query
    rows = cursor.fetchall()

    print(f"Total records retrieved: {len(rows)}\n")
    print("--- Displaying Data ---")

    # 6. Loop through the rows and print each record
    for row in rows:
        print(f"ID: {row[0]} | Name: {row[1]} | Email: {row[2]}")

except mysql.connector.Error as error:
    # Handle database errors
    print(f"Failed to read data from MySQL table: {error}")

finally:
    # 7. Close the cursor and connection to free up system resources
    cursor.close()
    connection.close()
    print("\nMySQL connection is closed.")
```
